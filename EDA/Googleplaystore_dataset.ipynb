```python
df[df['Reviews'] == '3.0M']

df = df[df['Reviews'] != '3.0M']
```

```python
#now convert to int datatype
df['Reviews']=df['Reviews'].astype(int)
```

```python
df.info()
```

```python
#convert sizes to either M or K
df['Size']=df['Size'].str.replace('M','000')
df['Size'] = df['Size'].str.replace('k', '', regex=False) # so now M is replced with 000 so converted to k
```

```python
df['Size'].astype(float)
#now we are getting error because there is "10839    Varies with device" which is not replaced
```

```python
import numpy as np

df['Size'] = df['Size'].replace('Varies with device', np.nan)
```

```python
df['Size']=df['Size'].astype(float)
df['Size'].head()
```

```python
df['Installs'].unique()
```

```python
#to convert to int datatype  we need to remove "+" symbol
car_to_remove=['+']
for item in car_to_remove:
   # Remove +
    df['Installs'] = df['Installs'].str.replace('+', '', regex=False)

# Remove commas
    df['Installs'] = df['Installs'].str.replace(',', '', regex=False)

# Convert to integer
df['Installs'] = df['Installs'].astype(int)

print(df['Installs'].head())
```
