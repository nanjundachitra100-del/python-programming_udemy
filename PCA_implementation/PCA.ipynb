```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn. datasets import load_breast_cancer
cancer_dataset=load_breast_cancer()
cancer_dataset.keys()
```

```python
print(cancer_dataset.DESCR)
```

```python
df = pd.DataFrame(
    cancer_dataset['data'],
    columns=cancer_dataset['feature_names']
)
df.head()
```

```python
#apply standardisation
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaler.fit(df)
```

```python
#transform this data into another container
scaled_data=scaler.transform(df)
scaled_data
```

```python
#applying PCA algorithm
from sklearn.decomposition import PCA
pca=PCA(n_components=2) ## if we did not give 2 then it would craete n no. of varience
data_pca=pca.fit_transform(scaled_data)
data_pca
```

```python
pca.explained_variance_ # as we gave n_components=2 we are getting 2 vareinces
```

```python
import matplotlib.pyplot as plt

plt.scatter(
    data_pca[:, 0],
    data_pca[:, 1],
    cmap='viridis'
)

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA - Breast Cancer Dataset")
plt.colorbar(label="Cancer Class")
plt.show()
```
