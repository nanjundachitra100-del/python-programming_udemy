```python
from sklearn.datasets import make_classification
X,y=make_classification(n_samples=1000,n_redundant=0,n_features=2,n_clusters_per_class=1,weights=[0.90],random_state=12)
#n_redundant specifies how many redundant (duplicate-like) features should be generated.
# like A AND B ,where C is another feature C=A+B,here C is reduntant feature
#n_clusters_per_class : how many groups should be made in each class.A  ***   B (=1), A  ***   *** B(=2) 
#weights[0.90]=majority =90%

import pandas as pd

df1=pd.DataFrame(X,columns=['f1','f2']) 
df2=pd.DataFrame(y,columns=['target'])
final_df=pd.concat([df1,df2],axis=1)
final_df.head()

#to see values counts ,how many 0's and 1's 
final_df['target'].value_counts()

import matplotlib.pyplot as plt

plt.scatter(final_df['f1'],final_df['f2'], c=final_df['target'] )
```
