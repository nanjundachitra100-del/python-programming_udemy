```python
import numpy as np

lst_marks = [30, 45, 67, 57, 48, 39, 57, 49, 67]

minimum, Q1, median, Q3, maximum = np.quantile(
    lst_marks,
    [0, 0.25, 0.50, 0.75, 1]
)
#Common quantiles
#Quantile (q)	Meaning
#0	Minimum value
#0.25	First Quartile (Q1)
0#.50	Median (Q2)
#0.75	Third Quartile (Q3)
#1 is Maximum value

print("Minimum :", minimum)
print("Q1      :", Q1)
print("Median  :", median)
print("Q3      :", Q3)
print("Maximum :", maximum)

IQR=Q3-Q1
print("IQR:",IQR)

lower_fence=Q1-1.5*(IQR)
upper_fence=Q3+1.5*(IQR)

print("lower_fence:",lower_fence)
print("upper_fence:",upper_fence)
#anything less than lower_fence value and greater than the upper_fence value is considered as the outlier
```

```python
import seaborn as sns
sns.boxplot(lst_marks)#without outlier
```
