```python
import pandas as pd 
from sklearn.preprocessing import OneHotEncoder
df=pd.DataFrame({'colors':
                 ["red","blue","green","orange","yellow","black"]})
print(df.head())
```

```python
from sklearn.preprocessing import OneHotEncoder

# Create an object of OneHotEncoder
encoder = OneHotEncoder()

# Apply fit_transform
encoded = encoder.fit_transform(df[['colors']]).toarray()
#this would create an sparx matrix

print(encoded)
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

df = pd.DataFrame({
    "color": ["Red", "Blue", "Green", "Red", "Blue"]
})

lbl_encoder = LabelEncoder()

df["color"] = lbl_encoder.fit_transform(df["color"])

print(df)
lbl_encoder.transform([['Blue']])
```

```python
#ordinary encoder
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

df = pd.DataFrame({
    'size': ['small', 'medium', 'big']
})
#creating a instance
encoder = OrdinalEncoder(categories=[['small', 'medium', 'big']])

encoded = encoder.fit_transform(df[['size']])

print(encoded)
```
