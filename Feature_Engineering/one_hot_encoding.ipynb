```python
import pandas as pd 
from sklearn.preprocessing import OneHotEncoder
df=pd.DataFrame({'colors':
                 ["red","blue","green","orange","yellow","black"]})
print(df.head())
```

```python
from sklearn.preprocessing import OneHotEncoder

# Create an object of OneHotEncoder
encoder = OneHotEncoder()

# Apply fit_transform
encoded = encoder.fit_transform(df[['colors']]).toarray()
#this would create an sparx matrix

print(encoded)
```
