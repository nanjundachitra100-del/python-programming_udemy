```python
# ==========================================
# Handling Imbalanced Dataset
# Upsampling and Downsampling Example
# ==========================================

# Import required libraries
import numpy as np
import pandas as pd
from sklearn.utils import resample

# ------------------------------------------
# Step 1: Set a random seed
# ------------------------------------------
# This ensures that the random numbers generated
# are the same every time you run the code.
np.random.seed(123)

# ------------------------------------------
# Step 2: Define total number of samples
# ------------------------------------------
n_samples = 1000

# 90% of the data belongs to Class 0
class_0_ratio = 0.9

# Number of samples in each class
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0

print("Number of Class 0 samples:", n_class_0)
print("Number of Class 1 samples:", n_class_1)

# ------------------------------------------
# Step 3: Create Majority Class (Class 0)
# ------------------------------------------
# np.random.normal(loc, scale, size)
#
# loc   = Mean of the data
# scale = Standard deviation
# size  = Number of values to generate

class_0 = pd.DataFrame({
    "feature_1": np.random.normal(loc=0, scale=1, size=n_class_0),
    "feature_2": np.random.normal(loc=0, scale=1, size=n_class_0),
    "target": [0] * n_class_0
})

# ------------------------------------------
# Step 4: Create Minority Class (Class 1)
# ------------------------------------------

class_1 = pd.DataFrame({
    "feature_1": np.random.normal(loc=2, scale=1, size=n_class_1),
    "feature_2": np.random.normal(loc=2, scale=1, size=n_class_1),
    "target": [1] * n_class_1
})

# ------------------------------------------
# Step 5: Combine both classes
# ------------------------------------------

df = pd.concat([class_0, class_1]).reset_index(drop=True)

print("\nOriginal Dataset")
print(df.head())

print("\nClass Distribution")
print(df["target"].value_counts())

# ==========================================
# UPSAMPLING
# ==========================================

print("\n==============================")
print("UPSAMPLING")
print("==============================")

# Increase the minority class from 100 rows to 900 rows

class1_up = resample(
    class_1,
    replace=True,          # Allow duplicate rows
    n_samples=len(class_0),# Make minority class equal to majority class
    random_state=42        # Produce same result every run
)

# Combine original majority class with upsampled minority class
df_up = pd.concat([class_0, class1_up]).reset_index(drop=True)

print("\nAfter Upsampling")
print(df_up["target"].value_counts())

# ==========================================
# DOWNSAMPLING
# ==========================================

print("\n==============================")
print("DOWNSAMPLING")
print("==============================")

# Reduce majority class from 900 rows to 100 rows

class0_down = resample(
    class_0,
    replace=False,         # Don't allow duplicates
    n_samples=len(class_1),# Reduce to minority class size
    random_state=42
)

# Combine downsampled majority class with original minority class
df_down = pd.concat([class0_down, class_1]).reset_index(drop=True)

print("\nAfter Downsampling")
print(df_down["target"].value_counts())

# ==========================================
# Display first few rows
# ==========================================

print("\nUpsampled Dataset")
print(df_up.head())

print("\nDownsampled Dataset")
print(df_down.head())
```
