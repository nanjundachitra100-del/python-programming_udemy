```python
students_marks=(85, 90, 78, 92, 88)
print(students_marks)

students_marks=list(students_marks)
students_marks.append(99)
students_marks.extend([92,96])

print(students_marks)
average=sum(students_marks)/len(students_marks)

print("average marks:",average)

highest_marks=max(students_marks)
lowest_marks=min(students_marks)
print(f"{highest_marks} is the highest marks")
print(f"{lowest_marks} is the lowest marks")
```
